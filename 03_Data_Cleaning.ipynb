# Date Format Conversion

The Order Date and Ship Date columns are currently stored as object data types. They will be converted to datetime format to enable time-based sales analysis.

```python
df["Order Date"] = pd.to_datetime(df["Order Date"])
df["Ship Date"] = pd.to_datetime(df["Ship Date"])
```

```python
df[["Order Date", "Ship Date"]].dtypes
```

```python
df.info()
```

#  Feature Engineering

In this section, new time-based features will be created from the Order Date column to support monthly, quarterly, and yearly sales analysis.

```python
df["Year"] = df["Order Date"].dt.year
```

```python
df["Quarter"] = df["Order Date"].dt.quarter
```

```python
df["Month"] = df["Order Date"].dt.month
```

```python
df["Month Name"] = df["Order Date"].dt.month_name()
```
