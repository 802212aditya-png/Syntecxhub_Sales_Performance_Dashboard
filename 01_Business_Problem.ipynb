# Sales Performance & Business Insights Dashboard





# Project Introduction

This project focuses on analyzing sales transaction data to understand overall business performance and generate meaningful, data-driven business insights.

The project involves importing and cleaning raw sales data, performing exploratory data analysis, analyzing sales trends across monthly, quarterly, and yearly periods, identifying top-selling and low-performing products, and comparing sales performance across different regions and product categories.

Key business performance indicators such as total sales, total profit, order volume, growth rate, and other relevant KPIs will be calculated to evaluate the company's sales and profitability performance.

The analyzed data will then be transformed into an interactive Power BI dashboard that enables users to explore sales performance through visualizations, KPIs, filters, and comparative analysis.

The ultimate goal of this project is to convert raw sales data into actionable business insights that can help management understand performance, identify opportunities, recognize underperforming areas, and support better data-driven decision-making.

# Project Focus

- Data Cleaning and Preprocessing
- Exploratory Data Analysis (EDA)
- Sales Trend Analysis
- Product Performance Analysis
- Category and Sub-Category Analysis
- Regional Performance Analysis
- Profitability Analysis
- KPI and Growth Analysis
- Interactive Power BI Dashboard
- Business Insights and Recommendations

# Tools & Technologies

Python, Pandas, NumPy, Matplotlib, Seaborn, Jupyter Notebook, Power BI, DAX, Git & GitHub






# Business Problem

The company needs to analyze its raw sales data to understand overall sales performance, identify sales trends, evaluate product and regional performance, and track key business performance indicators.

# Project Objectives

- Import and clean the raw sales dataset by removing null values and duplicates.
- Analyze monthly, quarterly, and yearly sales trends.
- Identify top-selling products and low-performing items.
- Compare sales performance across different regions and categories.
- Calculate key performance indicators such as total revenue, profit, and growth rate.
- Build an interactive Sales Performance Dashboard using Power BI.
