# Load Dataset

Sometimes a CSV file is not stored in UTF-8 encoding. Since Pandas uses UTF-8 by default, reading such a file may raise a UnicodeDecodeError.

To solve this problem, we specify the correct encoding (such as latin1 or cp1252) while reading the file.

Example:

df = pd.read_csv("superstore.csv", encoding="latin1")

```python
df=pd.read_csv("SampleSuperstore.csv",encoding="latin1")
```
